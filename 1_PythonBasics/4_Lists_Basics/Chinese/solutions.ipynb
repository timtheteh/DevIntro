# 列表基础练习

## 练习 1：创建列表与索引

创建一个名为 `foods` 的列表，包含 5 个你最喜欢的食物（字符串形式）。使用索引分别打印**第一个**元素和**最后一个**元素（不要直接打印整个列表）。

```python
foods = ["pizza", "sushi", "tacos", "pasta", "curry"]

print(foods[0])
print(foods[-1])
```

## 练习 2：正索引与负索引

给定 `colors = ["red", "orange", "yellow", "green", "blue", "purple"]`，使用正索引打印**第二个**元素，使用负索引打印**倒数第二个**元素。

```python
colors = ["red", "orange", "yellow", "green", "blue", "purple"]

print(colors[1])
print(colors[-2])
```

## 练习 3：切片（Slicing）

给定 `numbers = [3, 6, 9, 12, 15, 18, 21, 24]`，使用切片在一次 `print()` 调用中打印**前 3 个元素**，再在另一次 `print()` 调用中打印**最后 2 个元素**。

```python
numbers = [3, 6, 9, 12, 15, 18, 21, 24]

print(numbers[:3])
print(numbers[-2:])
```

## 练习 4：`append()` 与 `pop()`

创建一个名为 `cart` 的列表，包含 `["bread", "milk"]`。使用 `append()` 在末尾添加 `"eggs"`，然后打印该列表。接着使用不带参数的 `pop()` 删除最后一个元素，再次打印列表以确认它已被移除。

```python
cart = ["bread", "milk"]

cart.append("eggs")
print(cart)

cart.pop()
print(cart)
```

## 练习 5：`insert()` 与 `remove()`

给定 `queue = ["Alice", "Bob", "Diana"]`，使用 `insert()` 在索引 `2` 处插入 `"Charlie"`（使顺序变为 Alice、Bob、Charlie、Diana），然后打印列表。接着使用 `remove()` **按值**删除 `"Bob"`，再次打印列表。

```python
queue = ["Alice", "Bob", "Diana"]

queue.insert(2, "Charlie")
print(queue)

queue.remove("Bob")
print(queue)
```

## 练习 6：`len()` 与 `in` 运算符

给定 `inventory = ["sword", "shield", "potion", "map"]`，使用 `len()` 打印 `inventory` 中元素的数量。然后检查 `"potion"` 是否在 `inventory` 中并打印结果，再单独检查 `"bow"` 是否在 `inventory` 中并打印该结果（使用 `in` 运算符）。

```python
inventory = ["sword", "shield", "potion", "map"]

print(len(inventory))
print("potion" in inventory)
print("bow" in inventory)
```

## 练习 7：`extend()` 对比 `append()`

给定 `team_a = ["Sam", "Priya"]` 和 `team_b = ["Jordan", "Lee"]`，创建 `team_a` 的一个**副本（copy）**，命名为 `merged_extend`（例如使用 `team_a[:]`），然后使用 `extend()` 将 `team_b` 的所有成员逐个加入其中，打印 `merged_extend`。接着再创建一个副本 `merged_append`，改用 `append()` 将 `team_b` 加入其中，打印 `merged_append`，并观察结果有何不同（它包含一个嵌套列表，而不是四个独立的名字）。

```python
team_a = ["Sam", "Priya"]
team_b = ["Jordan", "Lee"]

merged_extend = team_a[:]
merged_extend.extend(team_b)
print(merged_extend)

merged_append = team_a[:]
merged_append.append(team_b)
print(merged_append)
```

## 练习 8：遍历列表与嵌套列表

第一部分：给定 `pets = ["cat", "dog", "parrot"]`，使用 `for` 循环打印每个元素**转换为大写**后的结果（在循环中使用字符串的 `.upper()` 方法）。

第二部分：给定嵌套列表 `seating = [["Amy", "Ben"], ["Cara", "Drew"]]`，打印整个第一个内部列表（`seating[0]`），然后使用两次链式索引只打印 `"Drew"`。

```python
pets = ["cat", "dog", "parrot"]

for pet in pets:
    print(pet.upper())

seating = [["Amy", "Ben"], ["Cara", "Drew"]]

print(seating[0])
print(seating[1][1])
```
