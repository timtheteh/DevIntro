# Lambda 基础练习

## 练习 1 — 零参数 Lambda编写一个**零参数**的 lambda，命名为 `greet`，返回固定字符串 `"Hello there!"`。将其赋值给一个变量，然后调用它并打印结果。

````python
greet = lambda: "Hello there!"
print(greet())
````

## 练习 2 — 单参数 Lambda编写一个名为 `square` 的 lambda，接收一个数字并返回它的平方。将其赋值给一个变量，然后用至少两个不同的数字调用它，并打印每个结果。

````python
square = lambda x: x * x
print(square(4))
print(square(9))
````

## 练习 3 — 立即调用 lambda（内联写法）编写一个接收一个数字并返回其两倍的 lambda——但这次不要先把它赋值给变量，而是**立即**定义并调用它，全部写在同一行（用括号把它包起来，然后立刻调用）。请对至少两个不同的输入数字这样做。

````python
print((lambda x: x * 2)(3))
print((lambda x: x * 2)(10))
````

## 练习 4 — 双参数 Lambda编写一个名为 `add` 的 lambda，接收两个数字并返回它们的和。将其赋值给一个变量，然后用两对不同的数字调用它，并打印每个结果。

````python
add = lambda a, b: a + b
print(add(2, 3))
print(add(10, 7))
````

## 练习 5 — 返回布尔值的双参数 Lambda编写一个名为 `is_greater` 的 lambda，接收两个数字 `a` 和 `b`，如果 `a` 大于 `b` 则返回 `True`，否则返回 `False`。用不同的数字对多次调用它，展示两种结果。

````python
is_greater = lambda a, b: a > b
print(is_greater(5, 3))
print(is_greater(2, 8))
print(is_greater(4, 4))
````

## 练习 6 — 多个（三个）参数编写一个名为 `box_volume` 的 lambda，接收三个参数——`length`（长）、`width`（宽）和 `height`（高）——并返回它们的乘积（也就是一个箱子的体积）。用至少两组不同的尺寸调用它。

````python
box_volume = lambda length, width, height: length * width * height
print(box_volume(2, 3, 4))
print(box_volume(5, 5, 5))
````

## 练习 7 — Lambda 中的条件表达式编写一个名为 `classify` 的 lambda，接收一个数字，如果大于 0 则返回字符串 `"positive"`，小于 0 则返回 `"negative"`，等于 0 则返回 `"zero"`。请记住：lambda 的函数体必须仍然是单一表达式，因此你需要用到 `x if condition else y` 这种条件表达式形式（可能需要链式使用）。分别用一个正数、一个负数和 0 调用它，展示全部三种结果。

````python
classify = lambda n: "positive" if n > 0 else ("negative" if n < 0 else "zero")
print(classify(5))
print(classify(-3))
print(classify(0))
````
