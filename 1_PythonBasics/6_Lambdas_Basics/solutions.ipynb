# Lambdas Basics Exercises

## Exercise 1 — Zero-argument lambda

Write a lambda with **zero arguments** called `greet` that returns the fixed string `"Hello there!"`. Assign it to a variable and then call it, printing the result.

````python
greet = lambda: "Hello there!"
print(greet())
````

## Exercise 2 — One-argument lambda

Write a lambda called `square` that takes one number and returns that number squared. Assign it to a variable, then call it with at least two different numbers and print each result.

````python
square = lambda x: x * x
print(square(4))
print(square(9))
````

## Exercise 3 — Calling a lambda immediately (inline)

Write a lambda that takes one number and returns it doubled — but instead of assigning it to a variable first, define it and call it **immediately, all in one line** (wrap it in parentheses and call it right away). Do this for at least two different input numbers.

````python
print((lambda x: x * 2)(3))
print((lambda x: x * 2)(10))
````

## Exercise 4 — Two-argument lambda

Write a lambda called `add` that takes two numbers and returns their sum. Assign it to a variable, then call it with two different pairs of numbers and print each result.

````python
add = lambda a, b: a + b
print(add(2, 3))
print(add(10, 7))
````

## Exercise 5 — Two-argument lambda returning a boolean

Write a lambda called `is_greater` that takes two numbers, `a` and `b`, and returns `True` if `a` is greater than `b`, otherwise `False`. Call it a few times with different pairs of numbers to show both outcomes.

````python
is_greater = lambda a, b: a > b
print(is_greater(5, 3))
print(is_greater(2, 8))
print(is_greater(4, 4))
````

## Exercise 6 — Multiple (three) arguments

Write a lambda called `box_volume` that takes three arguments — `length`, `width`, and `height` — and returns their product (the volume of a box). Call it with at least two different sets of dimensions.

````python
box_volume = lambda length, width, height: length * width * height
print(box_volume(2, 3, 4))
print(box_volume(5, 5, 5))
````

## Exercise 7 — Conditional expression inside a lambda

Write a lambda called `classify` that takes one number and returns the string `"positive"` if it's greater than 0, `"negative"` if it's less than 0, or `"zero"` if it equals 0. Remember: the lambda body must still be a single expression, so you'll need the `x if condition else y` conditional expression form (possibly chained). Call it with a positive number, a negative number, and zero to show all three results.

````python
classify = lambda n: "positive" if n > 0 else ("negative" if n < 0 else "zero")
print(classify(5))
print(classify(-3))
print(classify(0))
````
