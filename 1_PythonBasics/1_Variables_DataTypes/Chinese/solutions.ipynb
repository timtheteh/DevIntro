# 变量与数据类型练习

## 练习 1：创建并检查变量

创建五个变量，分别对应以下五种类型：`int`、`float`、`str`、`bool` 和 `None`。

然后使用 `print()` 和 `type()` 打印出每个变量的**值**和**类型**，每个变量一行。

```python
age = 25
price = 19.99
name = "Alice"
is_raining = True
result = None

print(age, type(age))
print(price, type(price))
print(name, type(name))
print(is_raining, type(is_raining))
print(result, type(result))
```

## 练习 2：算术运算符

创建两个 int 变量，`a = 17` 和 `b = 5`。

打印将每个算术运算符应用于 `a` 和 `b` 的结果：`+`、`-`、`*`、`/`、`//`、`%` 和 `**`。为每个结果加上标签，清楚地说明它是由哪个运算符产生的。

```python
a = 17
b = 5

print("a + b =", a + b)
print("a - b =", a - b)
print("a * b =", a * b)
print("a / b =", a / b)
print("a // b =", a // b)
print("a % b =", a % b)
print("a ** b =", a ** b)
```

## 练习 3：使用 input() 打招呼

使用 `input()` 询问用户的姓名，并将其存储在一个变量中。

再次使用 `input()` 询问用户的年龄，并将其存储在另一个变量中。

打印一条同时包含姓名和年龄的问候语（例如：`Hello Sam, you are 30 years old.`），然后打印年龄变量的 `type()`，以确认它是一个字符串。

```python
name = input("What is your name? ")
age = input("How old are you? ")

print("Hello " + name + ", you are " + age + " years old.")
print(type(age))
```

## 练习 4：将 input() 转换为数字

使用 `input()` 让用户输入一个数字。

使用 `int()` 将结果转换为 `int`，然后打印这个数字乘以 2 的结果，再打印这个数字加 10 的结果。同时打印转换后变量的 `type()`，以确认它现在是一个 int。

```python
number_text = input("Enter a number: ")
number = int(number_text)

print(number * 2)
print(number + 10)
print(type(number))
```

## 练习 5：比较两个数字

使用两次 `input()` 询问用户两个数字，并将每个数字都转换为 `float`。

打印使用每个比较运算符比较这两个数字的结果：`==`、`!=`、`<`、`>`、`<=` 和 `>=`。

```python
num1 = float(input("Enter the first number: "))
num2 = float(input("Enter the second number: "))

print("num1 == num2:", num1 == num2)
print("num1 != num2:", num1 != num2)
print("num1 < num2:", num1 < num2)
print("num1 > num2:", num1 > num2)
print("num1 <= num2:", num1 <= num2)
print("num1 >= num2:", num1 >= num2)
```

## 练习 6：逻辑运算符

创建两个 bool 变量：`has_ticket = True` 和 `is_on_time = False`。

打印 `has_ticket and is_on_time`、`has_ticket or is_on_time` 以及 `not has_ticket` 的结果，并为每个结果加上标签，清楚地说明每个结果代表的含义。

```python
has_ticket = True
is_on_time = False

print("has_ticket and is_on_time:", has_ticket and is_on_time)
print("has_ticket or is_on_time:", has_ticket or is_on_time)
print("not has_ticket:", not has_ticket)
```

## 练习 7：使用复合赋值运算符计算累计总量

创建一个变量 `total`，并将其设为 `100`。

仅使用复合赋值运算符（`+=`、`-=`、`*=`、`/=`），依次应用以下变化，并在每一步之后打印 `total`：
1. 加 50
2. 减 30
3. 乘以 2
4. 除以 4

```python
total = 100

total += 50
print("After += 50:", total)

total -= 30
print("After -= 30:", total)

total *= 2
print("After *= 2:", total)

total /= 4
print("After /= 4:", total)
```

## 练习 8：综合运用

使用 `input()` 询问用户两个数字，并将两者都转换为 `float`。

1. 打印这两个数字的和、差、积，以及商（`/`）。
2. 打印一个 bool 值，表示第一个数字是否大于第二个数字。
3. 创建一个变量 `both_positive`，使用逻辑运算符检查这两个数字是否**都**大于 0，并打印它。
4. 最后，创建一个从 `0` 开始的变量 `score`，并使用复合赋值运算符，对上面的每一项检查（大于比较和都为正数的检查）中结果为 `True` 的项给 `score` 加 1。打印最终的 `score`。

```python
num1 = float(input("Enter the first number: "))
num2 = float(input("Enter the second number: "))

print("Sum:", num1 + num2)
print("Difference:", num1 - num2)
print("Product:", num1 * num2)
print("Quotient:", num1 / num2)

is_first_greater = num1 > num2
print("First greater than second:", is_first_greater)

both_positive = num1 > 0 and num2 > 0
print("Both positive:", both_positive)

score = 0
score += is_first_greater
score += both_positive
print("Score:", score)
```
