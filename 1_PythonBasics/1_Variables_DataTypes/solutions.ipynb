## Exercise 1: Create and Inspect Variables

Create five variables, one of each of these types: `int`, `float`, `str`, `bool`, and `None`.

Then use `print()` and `type()` to print the **value** and the **type** of each variable, one pair per line.

```python
age = 25
price = 19.99
name = "Alice"
is_raining = True
result = None

print(age, type(age))
print(price, type(price))
print(name, type(name))
print(is_raining, type(is_raining))
print(result, type(result))
```

## Exercise 2: Arithmetic Operators

Create two int variables, `a = 17` and `b = 5`.

Print the result of each arithmetic operator applied to `a` and `b`: `+`, `-`, `*`, `/`, `//`, `%`, and `**`. Label each result so it is clear which operator produced it.

```python
a = 17
b = 5

print("a + b =", a + b)
print("a - b =", a - b)
print("a * b =", a * b)
print("a / b =", a / b)
print("a // b =", a // b)
print("a % b =", a % b)
print("a ** b =", a ** b)
```

## Exercise 3: Greeting with input()

Use `input()` to ask the user for their name, and store it in a variable.

Use `input()` again to ask the user for their age, and store it in another variable.

Print a greeting message that includes both the name and the age (e.g. `Hello Sam, you are 30 years old.`), then print the `type()` of the age variable to confirm it is a string.

## Exercise 4: Converting input() to a Number

Use `input()` to ask the user to enter a number.

Convert the result to an `int` using `int()`, then print the number doubled, and print the number plus 10. Also print the `type()` of the converted variable to confirm it is now an int.

```python
number_text = input("Enter a number: ")
number = int(number_text)

print(number * 2)
print(number + 10)
print(type(number))
```

## Exercise 5: Comparing Two Numbers

Use `input()` twice to ask the user for two numbers, converting each to a `float`.

Print the result of comparing the two numbers using each comparison operator: `==`, `!=`, `<`, `>`, `<=`, and `>=`.

```python
num1 = float(input("Enter the first number: "))
num2 = float(input("Enter the second number: "))

print("num1 == num2:", num1 == num2)
print("num1 != num2:", num1 != num2)
print("num1 < num2:", num1 < num2)
print("num1 > num2:", num1 > num2)
print("num1 <= num2:", num1 <= num2)
print("num1 >= num2:", num1 >= num2)
```

## Exercise 6: Logical Operators

Create two bool variables: `has_ticket = True` and `is_on_time = False`.

Print the result of `has_ticket and is_on_time`, `has_ticket or is_on_time`, and `not has_ticket`, with a label for each so it's clear what each result means.

```python
has_ticket = True
is_on_time = False

print("has_ticket and is_on_time:", has_ticket and is_on_time)
print("has_ticket or is_on_time:", has_ticket or is_on_time)
print("not has_ticket:", not has_ticket)
```

## Exercise 7: Running Total with Augmented Assignment

Create a variable `total` and set it to `100`.

Using only augmented assignment operators (`+=`, `-=`, `*=`, `/=`), apply the following changes one at a time, printing `total` after each step:
1. Add 50
2. Subtract 30
3. Multiply by 2
4. Divide by 4

```python
total = 100

total += 50
print("After += 50:", total)

total -= 30
print("After -= 30:", total)

total *= 2
print("After *= 2:", total)

total /= 4
print("After /= 4:", total)
```

## Exercise 8: Combining It All

Use `input()` to ask the user for two numbers, converting both to `float`.

1. Print the sum, difference, product, and quotient (`/`) of the two numbers.
2. Print a bool showing whether the first number is greater than the second number.
3. Create a variable `both_positive` that uses a logical operator to check whether **both** numbers are greater than 0, and print it.
4. Finally, create a variable `score` starting at `0`, and use augmented assignment to add 1 to `score` for every one of the checks above (greater-than check and both-positive check) that came out `True`. Print the final `score`.

```python
num1 = float(input("Enter the first number: "))
num2 = float(input("Enter the second number: "))

print("Sum:", num1 + num2)
print("Difference:", num1 - num2)
print("Product:", num1 * num2)
print("Quotient:", num1 / num2)

is_first_greater = num1 > num2
print("First greater than second:", is_first_greater)

both_positive = num1 > 0 and num2 > 0
print("Both positive:", both_positive)

score = 0
score += is_first_greater
score += both_positive
print("Score:", score)
```
