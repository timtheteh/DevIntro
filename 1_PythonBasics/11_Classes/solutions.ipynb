## Exercise 1

Define a class called `Book` with an `__init__` method that takes `title` and `author` as parameters and sets them as **instance attributes**. Then create one `Book` instance and print its `title` and `author`.

```python
class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author


book = Book("1984", "George Orwell")
print(book.title)
print(book.author)
```

## Exercise 2

Define a class called `Club` that has a **class attribute** `club_name = "Chess Club"` (shared by every instance) and an `__init__` method that sets an **instance attribute** `member_name`. Create two different `Club` instances with different `member_name` values, then print `club_name` and `member_name` for both to show that `club_name` is shared.

```python
class Club:
    club_name = "Chess Club"

    def __init__(self, member_name):
        self.member_name = member_name


member1 = Club("Alice")
member2 = Club("Bob")

print(member1.club_name, member1.member_name)
print(member2.club_name, member2.member_name)
```

## Exercise 3

Define a class called `Rectangle` with an `__init__` method that sets `width` and `height` as instance attributes. Add a `__str__` method that returns a readable string such as `"Rectangle(3 x 4)"`. Create an instance and use `print()` on it to confirm `__str__` is being called.

```python
class Rectangle:
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def __str__(self):
        return f"Rectangle({self.width} x {self.height})"


rect = Rectangle(3, 4)
print(rect)
```

## Exercise 4

Take the `Rectangle` class from Exercise 3 and add a `__repr__` method that returns an unambiguous string such as `"Rectangle(width=3, height=4)"`. Create an instance and call `repr()` on it directly, then put the instance inside a list and print the list to see `__repr__` used automatically for the list's element.

```python
class Rectangle:
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def __str__(self):
        return f"Rectangle({self.width} x {self.height})"

    def __repr__(self):
        return f"Rectangle(width={self.width}, height={self.height})"


rect = Rectangle(3, 4)
print(repr(rect))

rectangles = [rect]
print(rectangles)
```

## Exercise 5

Add an `__eq__` method to the `Rectangle` class so that two `Rectangle` instances are considered equal (`==`) if they have the same `width` and same `height`. Create two separate `Rectangle` instances with identical dimensions and confirm `==` returns `True`, then create a third with different dimensions and confirm `==` returns `False`.

```python
class Rectangle:
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def __str__(self):
        return f"Rectangle({self.width} x {self.height})"

    def __repr__(self):
        return f"Rectangle(width={self.width}, height={self.height})"

    def __eq__(self, other):
        return self.width == other.width and self.height == other.height


rect_a = Rectangle(3, 4)
rect_b = Rectangle(3, 4)
rect_c = Rectangle(5, 6)

print(rect_a == rect_b)
print(rect_a == rect_c)
```

## Exercise 6

Define a subclass called `Square` that inherits from `Rectangle`. Its `__init__` should take a single parameter `side` and call `super().__init__(...)` to set both `width` and `height` to `side`. Create a `Square` instance and print it (this should use the `__str__` method inherited from `Rectangle`).

```python
class Square(Rectangle):
    def __init__(self, side):
        super().__init__(side, side)


square = Square(5)
print(square)
```

## Exercise 7

Define a base class `Vehicle` with an `__init__` method that sets `make` and `model` as instance attributes. Define a subclass `Car` whose `__init__` takes `make`, `model`, and `num_doors`; it should call `super().__init__(make, model)` and then set `num_doors` as its own instance attribute. Create a `Car` instance and print all three attributes (`make`, `model`, `num_doors`).

```python
class Vehicle:
    def __init__(self, make, model):
        self.make = make
        self.model = model


class Car(Vehicle):
    def __init__(self, make, model, num_doors):
        super().__init__(make, model)
        self.num_doors = num_doors


car = Car("Toyota", "Corolla", 4)
print(car.make)
print(car.model)
print(car.num_doors)
```

## Exercise 8

Define a class `Person` with:
- a class attribute `species = "Homo sapiens"`
- an `__init__` that sets instance attributes `name` and `age`
- a `__str__` method returning something like `"Alice, age 30"`
- a `__repr__` method returning something like `"Person(name='Alice', age=30)"`
- an `__eq__` method that considers two `Person` instances equal if `name` and `age` both match

Then define a subclass `Student` that inherits from `Person`. Its `__init__` should take `name`, `age`, and `school`; it should call `super().__init__(name, age)` and then set `school` as its own instance attribute. Create a `Student` instance and print it, print its `repr()`, and print its `school`.

```python
class Person:
    species = "Homo sapiens"

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __str__(self):
        return f"{self.name}, age {self.age}"

    def __repr__(self):
        return f"Person(name={self.name!r}, age={self.age!r})"

    def __eq__(self, other):
        return self.name == other.name and self.age == other.age


class Student(Person):
    def __init__(self, name, age, school):
        super().__init__(name, age)
        self.school = school


student = Student("Alice", 20, "State University")
print(student)
print(repr(student))
print(student.school)
```
