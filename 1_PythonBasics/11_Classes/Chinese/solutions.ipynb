# 类（Classes）语法练习

## 练习 1

定义一个名为 `Book` 的类，其 `__init__` 方法接收 `title` 和 `author` 两个参数，并将它们设置为**实例属性**。然后创建一个 `Book` 实例，并打印它的 `title` 和 `author`。

```python
class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author


book = Book("1984", "George Orwell")
print(book.title)
print(book.author)
```

## 练习 2

定义一个名为 `Club` 的类，它拥有一个**类属性** `club_name = "Chess Club"`（被每一个实例共享），以及一个用于设置**实例属性** `member_name` 的 `__init__` 方法。创建两个 `member_name` 值不同的 `Club` 实例，然后分别打印两者的 `club_name` 和 `member_name`，以展示 `club_name` 是被共享的。

```python
class Club:
    club_name = "Chess Club"

    def __init__(self, member_name):
        self.member_name = member_name


member1 = Club("Alice")
member2 = Club("Bob")

print(member1.club_name, member1.member_name)
print(member2.club_name, member2.member_name)
```

## 练习 3

定义一个名为 `Rectangle` 的类，其 `__init__` 方法将 `width` 和 `height` 设置为实例属性。添加一个 `__str__` 方法，返回一个可读的字符串，例如 `"Rectangle(3 x 4)"`。创建一个实例并对其使用 `print()`，以确认 `__str__` 被调用了。

```python
class Rectangle:
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def __str__(self):
        return f"Rectangle({self.width} x {self.height})"


rect = Rectangle(3, 4)
print(rect)
```

## 练习 4

在练习 3 的 `Rectangle` 类基础上，添加一个 `__repr__` 方法，返回一个明确无歧义的字符串，例如 `"Rectangle(width=3, height=4)"`。创建一个实例，直接对它调用 `repr()`，然后把该实例放进一个列表中并打印这个列表，以观察 `__repr__` 是如何被自动用于列表元素的。

```python
class Rectangle:
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def __str__(self):
        return f"Rectangle({self.width} x {self.height})"

    def __repr__(self):
        return f"Rectangle(width={self.width}, height={self.height})"


rect = Rectangle(3, 4)
print(repr(rect))

rectangles = [rect]
print(rectangles)
```

## 练习 5

为 `Rectangle` 类添加一个 `__eq__` 方法，使得当两个 `Rectangle` 实例拥有相同的 `width` 和相同的 `height` 时，被视为相等（`==`）。创建两个尺寸相同的独立 `Rectangle` 实例，确认 `==` 返回 `True`；再创建一个尺寸不同的第三个实例，确认 `==` 返回 `False`。

```python
class Rectangle:
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def __str__(self):
        return f"Rectangle({self.width} x {self.height})"

    def __repr__(self):
        return f"Rectangle(width={self.width}, height={self.height})"

    def __eq__(self, other):
        return self.width == other.width and self.height == other.height


rect_a = Rectangle(3, 4)
rect_b = Rectangle(3, 4)
rect_c = Rectangle(5, 6)

print(rect_a == rect_b)
print(rect_a == rect_c)
```

## 练习 6

定义一个名为 `Square` 的子类，继承自 `Rectangle`。它的 `__init__` 应接收一个单独的参数 `side`，并调用 `super().__init__(...)` 将 `width` 和 `height` 都设置为 `side`。创建一个 `Square` 实例并打印它（这应该会使用从 `Rectangle` 继承来的 `__str__` 方法）。

```python
class Square(Rectangle):
    def __init__(self, side):
        super().__init__(side, side)


square = Square(5)
print(square)
```

## 练习 7

定义一个基类 `Vehicle`，其 `__init__` 方法将 `make` 和 `model` 设置为实例属性。定义一个子类 `Car`，其 `__init__` 接收 `make`、`model` 和 `num_doors`；它应调用 `super().__init__(make, model)`，然后将 `num_doors` 设置为自己的实例属性。创建一个 `Car` 实例，并打印全部三个属性（`make`、`model`、`num_doors`）。

```python
class Vehicle:
    def __init__(self, make, model):
        self.make = make
        self.model = model


class Car(Vehicle):
    def __init__(self, make, model, num_doors):
        super().__init__(make, model)
        self.num_doors = num_doors


car = Car("Toyota", "Corolla", 4)
print(car.make)
print(car.model)
print(car.num_doors)
```

## 练习 8

定义一个类 `Person`，包含：
- 一个类属性 `species = "Homo sapiens"`
- 一个用于设置实例属性 `name` 和 `age` 的 `__init__`
- 一个返回类似 `"Alice, age 30"` 的 `__str__` 方法
- 一个返回类似 `"Person(name='Alice', age=30)"` 的 `__repr__` 方法
- 一个 `__eq__` 方法，当 `name` 和 `age` 都匹配时，认为两个 `Person` 实例相等

然后定义一个继承自 `Person` 的子类 `Student`。它的 `__init__` 应接收 `name`、`age` 和 `school`；它应调用 `super().__init__(name, age)`，然后将 `school` 设置为自己的实例属性。创建一个 `Student` 实例，打印它、打印它的 `repr()`，并打印它的 `school`。

```python
class Person:
    species = "Homo sapiens"

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __str__(self):
        return f"{self.name}, age {self.age}"

    def __repr__(self):
        return f"Person(name={self.name!r}, age={self.age!r})"

    def __eq__(self, other):
        return self.name == other.name and self.age == other.age


class Student(Person):
    def __init__(self, name, age, school):
        super().__init__(name, age)
        self.school = school


student = Student("Alice", 20, "State University")
print(student)
print(repr(student))
print(student.school)
```
