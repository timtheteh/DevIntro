## Exercise 2: Concatenation with `+`

Create two variables, `first_name` and `last_name`, each holding a name of your choice. Use `+` to combine them into a single string called `full_name`, with a space in between, and print the result.

```python
first_name = "Ada"
last_name = "Lovelace"
full_name = first_name + " " + last_name
print(full_name)
```

## Exercise 3: Repetition with `*`

Create a variable called `divider` that uses `*` to repeat the character `"="` 20 times (without typing it out 20 times manually). Print `divider`.

```python
divider = "=" * 20
print(divider)
```

## Exercise 4: Indexing (Positive and Negative)

## Exercise 5: Slicing

## Exercise 7: Cleaning Up a Messy String

Given the variable `messy = "   HELLO there   "`, use string methods to:
1. Remove the extra whitespace from the start and end (`.strip()`)
2. Convert the result to lowercase (`.lower()`)

Store the final cleaned-up string in a variable called `clean`, and print it. (Tip: you can chain methods one after another, e.g. `messy.strip().lower()`.)

```python
messy = "   HELLO there   "

clean = messy.strip().lower()
print(clean)
```

## Exercise 8: Splitting, Joining, Checking, and Converting

Given the variable `sentence = "the quick brown fox jumps"`, do the following:
1. Use `.split()` to break `sentence` into a list of words, stored in a variable called `words`.
2. Use the `in` operator to check whether the word `"fox"` appears in `sentence`, and print `True`/`False`.
3. Use `.replace()` on `sentence` to change `"fox"` to `"cat"`, storing the result in `new_sentence`, and print it.
4. Create a variable `count = 5`, and print a message like `"There are 5 words"` by concatenating a string with `str(count)` using `+`.

```python
sentence = "the quick brown fox jumps"
count = 5

words = sentence.split()
print(words)

print("fox" in sentence)

new_sentence = sentence.replace("fox", "cat")
print(new_sentence)

print("There are " + str(count) + " words")
```
