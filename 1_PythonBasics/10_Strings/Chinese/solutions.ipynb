# 字符串语法练习

## 练习 1 — 基本 f-字符串

给定变量 `city = "Paris"` 和 `population = 2148000`，使用 f-字符串打印：

`Paris has a population of 2148000`

（只使用 f-字符串构建这句话——不要用 `+` 拼接。）

````python
city = "Paris"
population = 2148000

print(f"{city} has a population of {population}")
````

## 练习 2 — 格式说明符：固定小数位数

给定 `price = 19.6`，使用带格式说明符的 f-字符串将其打印为恰好 **2 位小数**，例如 `19.60`。

````python
price = 19.6

print(f"{price:.2f}")
````

## 练习 3 — 调试说明符 `=`

给定 `radius = 7`，使用 f-字符串的调试说明符（`f"{radius=}"`）在一行中同时打印表达式文本及其值，不要自行添加额外的打印格式。

````python
radius = 7

print(f"{radius=}")
````

## 练习 4 — 转换标志 `!r`

给定 `word = "hello\n"`，使用带 `!r` 转换标志的 f-字符串打印 `word` 的 **repr**（这样转义序列会显示为 `\n`，而不是真正的换行）。

````python
word = "hello\n"

print(f"{word!r}")
````

## 练习 5 — 使用位置和关键字占位符的 `.format()`

使用 `.format()` 方法（而不是 f-字符串），生成字符串：

`Order #42: 3 x Widget`

要求：
- 对订单号使用**显式位置索引**（例如 `{0}`），以便它可以在模板中的其他位置被复用。
- 对 `qty` 和 `product` 使用**关键字参数**。

````python
order_number = 42
qty = 3
product = "Widget"

result = "Order #{0}: {qty} x {product}".format(order_number, qty=qty, product=product)
print(result)
````

## 练习 6 — 用于 Windows 路径的原始字符串

编写一个**原始字符串字面量**（raw string literal）来正确表示以下 Windows 文件路径，然后将其打印出来：

`C:\Users\Ada\Documents\notes.txt`

不要使用任何转义序列——改用 `r"..."` 前缀。

````python
path = r"C:\Users\Ada\Documents\notes.txt"
print(path)
````

## 练习 8 — 综合运用格式说明符、转义序列和 `.split()`

1. 使用真正的转义序列（而不是原始字符串）创建一个等于 `"Name\tScore"` 的字符串变量 `row`。
2. 打印 `row`（应显示为两个以制表符分隔的单词）。
3. 给定 `score = 87.456`，使用 f-字符串打印 `Formatted score: 87.46`（2 位小数），**并且**在同一个 f-字符串内换行（使用 `\n`），将同一个值打印为补零的、宽度为 6 位字符的整数字段，例如 `000087`。
4. 最后，取字符串 `"apple,banana,,cherry"`，按逗号拆分，并使用 `.join()` 只将**非空**部分重新组合成一个逗号分隔的字符串（这是对 `.split()` + `.join()` 的语法练习——用简单的列表推导式或过滤方式即可）。

````python
score = 87.456
raw_data = "apple,banana,,cherry"

row = "Name\tScore"
print(row)

print(f"Formatted score: {score:.2f}\nZero-padded: {score:06.0f}")

parts = raw_data.split(",")
non_empty = [p for p in parts if p]
joined = ",".join(non_empty)
print(joined)
````
