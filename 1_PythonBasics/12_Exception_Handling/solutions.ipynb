# Exception Handling Exercises

## Exercise 1: Basic try/except

Write a `try`/`except` block that attempts to compute `10 / 0` and catches the resulting `ZeroDivisionError`, printing the message `"Cannot divide by zero!"` when it happens.

```python
try:
    result = 10 / 0
except ZeroDivisionError:
    print("Cannot divide by zero!")
```

## Exercise 2: Catching a specific exception and reading its message

Write a `try`/`except` block that attempts to run `int("abc")`. Catch the `ValueError` using `as e`, and print the exception object `e` so its error message is displayed.

```python
try:
    number = int("abc")
except ValueError as e:
    print(e)
```

## Exercise 3: Multiple except clauses

Write a `try`/`except` block with **two separate `except` clauses**: one for `ValueError` and one for `TypeError`. Inside the `try`, put code that could raise either (e.g. converting a value with `int(...)`, or adding incompatible types). Each `except` clause should print a different, distinct message.

```python
try:
    result = "5" + 5
except ValueError:
    print("A ValueError occurred.")
except TypeError:
    print("A TypeError occurred.")
```

## Exercise 4: Catching multiple exception types in one clause

Write a single `try`/`except` block where the `except` clause catches **both** `ValueError` and `ZeroDivisionError` using a tuple, e.g. `except (ValueError, ZeroDivisionError) as e:`. Print the caught exception's message.

```python
try:
    value = int("abc")
    result = 100 / value
except (ValueError, ZeroDivisionError) as e:
    print(f"Something went wrong: {e}")
```

## Exercise 5: Full try/except/else/finally

Write a `try` block that converts the string `"42"` to an integer. Include:
- an `except ValueError` clause,
- an `else` clause that prints the converted number (this should run since no error occurs),
- a `finally` clause that always prints `"Done."`.

```python
try:
    number = int("42")
except ValueError:
    print("Conversion failed.")
else:
    print(number)
finally:
    print("Done.")
```

## Exercise 6: Raising a built-in exception

Write a function `check_positive(number)` that raises a `ValueError` with the message `"Number must be positive"` if `number` is less than 0, and otherwise prints `"Number is valid"`. Then call it once with a negative number inside a `try`/`except` block that catches and prints the error.

```python
def check_positive(number):
    if number < 0:
        raise ValueError("Number must be positive")
    print("Number is valid")

try:
    check_positive(-5)
except ValueError as e:
    print(e)
```

## Exercise 7: Re-raising an exception

Write a `try`/`except` block around `1 / 0`. In the `except ZeroDivisionError` clause, first print `"Logging error before re-raising..."`, then use a bare `raise` statement to re-raise the same exception. Wrap the whole thing in an outer `try`/`except` so the re-raised exception is caught and its message is printed (you should see both print statements).

```python
try:
    try:
        1 / 0
    except ZeroDivisionError:
        print("Logging error before re-raising...")
        raise
except ZeroDivisionError as e:
    print(e)
```

## Exercise 8: Custom exception class

Define a custom exception class called `InvalidAgeError` that subclasses `Exception`. Then write a function `set_age(age)` that raises `InvalidAgeError` with the message `"Age cannot be negative"` if `age < 0`, and otherwise prints `f"Age set to {age}"`. Call the function with a negative age inside a `try`/`except InvalidAgeError` block that catches and prints the custom error.

```python
class InvalidAgeError(Exception):
    """Raised when an age value is invalid."""
    pass

def set_age(age):
    if age < 0:
        raise InvalidAgeError("Age cannot be negative")
    print(f"Age set to {age}")

try:
    set_age(-1)
except InvalidAgeError as e:
    print(f"Custom error caught: {e}")
```
