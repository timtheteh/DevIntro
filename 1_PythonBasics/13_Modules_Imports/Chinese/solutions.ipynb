# 模块与导入（Modules & Imports）练习

## 练习 1 —— 基础 `import`

使用 `import` 导入整个 `math` 模块，然后使用它（带上 `math.` 前缀）打印 `144` 的平方根。

```python
import math

print(math.sqrt(144))
```

## 练习 2 —— `from module import name`

使用 `from ... import ...` 语法，**只** 导入 `random` 模块中的 `choice` 函数。然后直接使用它（不加模块前缀）从列表 `["rock", "paper", "scissors"]` 中随机选取一个元素。

```python
from random import choice

options = ["rock", "paper", "scissors"]
pick = choice(options)
print(pick)
```

## 练习 3 —— 使用 `as` 为模块设置别名

导入 `datetime` 模块，**并将其别名设为 `dt`**。然后使用 `dt.datetime.now()` 获取当前日期和时间，并只打印结果的 `.year` 属性。

```python
import datetime as dt

now = dt.datetime.now()
print(now.year)
```

## 练习 4 —— 使用 `as` 为导入的名字设置别名

从 `os` 模块中 **只** 导入 `getcwd` 函数，并将其别名设为 `cwd_fn`（即 `from os import getcwd as cwd_fn`）。调用 `cwd_fn()` 并打印结果。

```python
from os import getcwd as cwd_fn

print(cwd_fn())
```

## 练习 5 —— 多个名字，一个设别名、一个不设

在一条 `from math import ...` 语句中，同时导入 `pi`（不设别名）**和** `floor`（别名设为 `round_down`）。直接打印 `pi`，然后打印调用 `round_down(pi)` 的结果。

```python
from math import pi, floor as round_down

print(pi)
print(round_down(pi))
```

## 练习 6 —— 两个模块，两种不同的写法

在同一个代码单元格中：`import json`（导入整个模块，不设别名），以及单独的 `from random import randint`（导入特定名字，不设别名）。使用 `randint(1, 100)` 生成一个随机数，将其放入一个像 `{"value": <该数字>}` 这样的字典中，然后使用 `json.dumps(...)` 将该字典打印为一个 JSON 字符串。

```python
import json
from random import randint

number = randint(1, 100)
data = {"value": number}
print(json.dumps(data))
```

## 练习 7 —— `__name__` 判断

编写一个名为 `describe_module()` 的函数，返回字符串 `"This code is running!"`。在函数定义下方，添加一个 `if __name__ == "__main__":` 代码块，调用 `describe_module()` 并打印其返回值。运行该单元格，确认它按预期打印结果——然后在下方用一条注释，用一句话解释：如果这段代码被保存到一个文件中，并从另一个脚本中被 *导入*（而不是直接运行），那条打印语句会发生什么。

```python
def describe_module():
    return "This code is running!"

if __name__ == "__main__":
    print(describe_module())

# 如果这段代码被保存到一个文件中，并从另一个脚本中被导入而不是直接运行，
# __name__ 的值将会是该模块自己的名字（而不是 "__main__"），因此这个 if 代码块——以及其中的打印语句——将会被跳过。
```

## 练习 8 —— 综合运用

使用一条 `import` 语句和一条 `from ... import ... as ...` 语句，完成以下任务：
1. 正常导入整个 `math` 模块（不设别名）。
2. 从 `datetime` 中导入 `datetime`（该类），并将其别名设为 `Date`。

然后打印 `math.pow(2, 5)` 和 `Date.now().year`，以确认两个导入都正常工作。

```python
import math
from datetime import datetime as Date

print(math.pow(2, 5))
print(Date.now().year)
```
