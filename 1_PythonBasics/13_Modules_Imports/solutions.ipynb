# Modules & Imports Exercises

## Exercise 1 — Basic `import`

Use `import` to bring in the whole `math` module, then use it (with the `math.` prefix) to print the square root of `144`.

```python
import math

print(math.sqrt(144))
```

## Exercise 2 — `from module import name`

Import **only** the `choice` function from the `random` module using `from ... import ...` syntax. Then use it directly (no module prefix) to pick a random item from the list `["rock", "paper", "scissors"]`.

```python
from random import choice

options = ["rock", "paper", "scissors"]
pick = choice(options)
print(pick)
```

## Exercise 3 — Aliasing a module with `as`

Import the `datetime` module **with the alias `dt`**. Then use `dt.datetime.now()` to get the current date and time and print just the `.year` attribute of the result.

```python
import datetime as dt

now = dt.datetime.now()
print(now.year)
```

## Exercise 4 — Aliasing an imported name with `as`

From the `os` module, import **only** the `getcwd` function, giving it the alias `cwd_fn` (i.e. `from os import getcwd as cwd_fn`). Call `cwd_fn()` and print the result.

```python
from os import getcwd as cwd_fn

print(cwd_fn())
```

## Exercise 5 — Multiple names, one aliased, one not

In a single `from math import ...` statement, import `pi` (no alias) **and** `floor` with the alias `round_down`. Print `pi` as-is, then print the result of calling `round_down(pi)`.

```python
from math import pi, floor as round_down

print(pi)
print(round_down(pi))
```

## Exercise 6 — Two modules, two different styles

In the same code cell: `import json` (whole module, no alias), and separately `from random import randint` (specific name, no alias). Use `randint(1, 100)` to generate a random number, put it in a dictionary like `{"value": <the number>}`, then use `json.dumps(...)` to print that dictionary as a JSON string.

```python
import json
from random import randint

number = randint(1, 100)
data = {"value": number}
print(json.dumps(data))
```

## Exercise 7 — The `__name__` guard

Write a function called `describe_module()` that returns the string `"This code is running!"`. Below the function definition, add an `if __name__ == "__main__":` block that calls `describe_module()` and prints its return value. Run the cell and confirm it prints as expected — then, in a comment underneath, explain in one sentence what would happen to that print statement if this code were saved to a file and *imported* from another script instead of run directly.

```python
def describe_module():
    return "This code is running!"

if __name__ == "__main__":
    print(describe_module())

# If this code were saved to a file and imported from another script instead of run directly,
# __name__ would equal the module's own name (not "__main__"), so the if-block — and its print — would be skipped.
```

## Exercise 8 — Mixing it all together

Using one `import` statement and one `from ... import ... as ...` statement, do the following:
1. Import the whole `math` module normally (no alias).
2. From `datetime`, import `datetime` (the class) with the alias `Date`.

Then print `math.pow(2, 5)` and `Date.now().year` to confirm both imports work.

```python
import math
from datetime import datetime as Date

print(math.pow(2, 5))
print(Date.now().year)
```
