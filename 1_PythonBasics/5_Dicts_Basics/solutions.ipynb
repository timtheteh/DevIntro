# Dicts Basics Exercises

## Exercise 1: Create a dict and access a value

Create a dict called `ages` that maps 3 names of your choice to their ages (e.g. `"Alice": 30`). Then print the age of just one of the names, using `ages[name]`.

```python
ages = {"Alice": 30, "Bob": 25, "Charlie": 35}

print(ages["Alice"])
```

## Exercise 2: `.get()` with a default value

Using the `ages` dict from Exercise 1, use `.get()` to look up a name that is **not** in the dict, passing `0` as the default value, and print the result. It should print `0` instead of crashing with a `KeyError`.

```python
print(ages.get("Diana", 0))
```

## Exercise 3: Add a new key and update an existing key

Using the `ages` dict, add a brand new name/age pair to it, then update the age of one of the names that was already in the dict. Print the dict after each change so you can see it grow and update.

```python
ages["Diana"] = 28
print(ages)

ages["Bob"] = 26
print(ages)
```

## Exercise 4: Remove a key with `del` and with `.pop()`

Using the `ages` dict, remove one of the names with the `del` statement. Then remove a different name using `.pop()`, storing the value it returns in a variable and printing that variable. Print the dict at the end to confirm both names are gone.

```python
del ages["Diana"]

removed_value = ages.pop("Charlie")
print(removed_value)

print(ages)
```

## Exercise 5: Check key existence with `in`

Using the `ages` dict, write an `if` statement that checks whether `"Bob"` is a key in the dict using the `in` operator. If it is, print its value; otherwise, print `"Bob not found"`.

```python
if "Bob" in ages:
    print(ages["Bob"])
else:
    print("Bob not found")
```

## Exercise 6: Loop over the keys (default iteration)

Create a dict called `capitals` that maps 3 country names to their capital cities. Write a `for` loop that iterates directly over `capitals` (no `.keys()`, `.values()`, or `.items()`) and prints each country name.

```python
capitals = {"France": "Paris", "Japan": "Tokyo", "Egypt": "Cairo"}

for country in capitals:
    print(country)
```

## Exercise 7: Loop over `.values()`

Using the `capitals` dict from Exercise 6, write a `for` loop using `.values()` that prints each capital city on its own line.

```python
for city in capitals.values():
    print(city)
```

## Exercise 8: Loop over `.items()`

Using the `capitals` dict from Exercise 6, write a `for` loop using `.items()` that prints each pair in the form `country: capital` (e.g. `France: Paris`).

```python
for country, city in capitals.items():
    print(country + ": " + city)
```
