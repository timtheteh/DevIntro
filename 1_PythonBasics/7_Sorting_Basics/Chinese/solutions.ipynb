# 排序基础练习

## 练习 1：对数字列表使用 `sorted()`

给定 `numbers = [8, 3, 10, 1, 6]`，使用 `sorted()` 创建一个新变量 `ascending`，保存升序排列的数字。分别打印 `numbers`（确认它没有被改变）和 `ascending`。

````python
numbers = [8, 3, 10, 1, 6]

ascending = sorted(numbers)

print(numbers)
print(ascending)
````

## 练习 2：使用 `.sort()` 原地排序

给定 `scores = [55, 92, 78, 40, 61]`，使用 `.sort()` 方法将列表原地排序（升序）。排序后打印 `scores`，确认它被直接修改了。

````python
scores = [55, 92, 78, 40, 61]

scores.sort()

print(scores)
````

## 练习 3：反向字母顺序

给定 `names = ["Priya", "Amit", "Wendy", "Carlos"]`，使用带有 `reverse=True` 的 `sorted()` 创建一个新列表 `names_desc`，其中包含按反向字母顺序（从 Z 到 A）排列的姓名。

````python
names = ["Priya", "Amit", "Wendy", "Carlos"]

names_desc = sorted(names, reverse=True)

print(names_desc)
````

## 练习 4：使用 `key=` 按字符串长度排序

给定 `words = ["kiwi", "banana", "fig", "watermelon", "pear"]`，使用带有 `key=lambda` 参数的 `sorted()`，将单词从最短到最长排序。将结果存入 `by_length`。

````python
words = ["kiwi", "banana", "fig", "watermelon", "pear"]

by_length = sorted(words, key=lambda word: len(word))

print(by_length)
````

## 练习 5：按元组的第二个元素排序

给定 `scores = [("Alice", 88), ("Bob", 95), ("Charlie", 72)]`，使用带有 `key=lambda` 的 `sorted()`，按每个人的分数（元组的第二个元素）对这个元组列表进行排序，从低到高。

````python
scores = [("Alice", 88), ("Bob", 95), ("Charlie", 72)]

by_score = sorted(scores, key=lambda pair: pair[1])

print(by_score)
````

## 练习 6：按某个键对字典列表排序

给定下方的列表，使用 `key=lambda d: d["age"]` 按每个人的 `"age"` 值升序排序。将结果存入 `by_age`，并逐行打印每个人的姓名和年龄。

```python
people = [
    {"name": "Alice", "age": 30},
    {"name": "Bob", "age": 25},
    {"name": "Charlie", "age": 35},
]
```

````python
people = [
    {"name": "Alice", "age": 30},
    {"name": "Bob", "age": 25},
    {"name": "Charlie", "age": 35},
]

by_age = sorted(people, key=lambda d: d["age"])

for person in by_age:
    print(person["name"], person["age"])
````

## 练习 7：对字典列表降序排序

使用练习 6 中相同的 `people` 列表，在一次 `sorted()` 调用中结合 `key=` 和 `reverse=True`，按 `"age"` 降序（年龄最大的在前）排序。将结果存入 `by_age_desc`。

````python
people = [
    {"name": "Alice", "age": 30},
    {"name": "Bob", "age": 25},
    {"name": "Charlie", "age": 35},
]

by_age_desc = sorted(people, key=lambda d: d["age"], reverse=True)

for person in by_age_desc:
    print(person["name"], person["age"])
````

## 练习 8：找出并修复 bug

下面这段代码的目的是按价格（最便宜的排在最前）对 `inventory`（一个字典列表，每个字典都有 `"item"` 和 `"price"` 两个键）进行排序，并将排序后的列表存入一个新变量 —— 但它有一个 bug：

```python
inventory = [
    {"item": "mug", "price": 12},
    {"item": "lamp", "price": 45},
    {"item": "pen", "price": 3},
]

cheapest_first = inventory.sort(key=lambda x: x["price"])
print(cheapest_first)
```

用注释解释这段代码错在哪里，然后在下面写出修正后的版本，使其生成一个新变量 `cheapest_first`，保存排序后的字典列表。

````python
inventory = [
    {"item": "mug", "price": 12},
    {"item": "lamp", "price": 45},
    {"item": "pen", "price": 3},
]

# Bug：.sort() 会原地排序列表并返回 None，
# 所以 cheapest_first 最终绑定的是 None，而不是排序后的列表。
# 应该改用 sorted()，因为它会返回一个新的排序列表。

cheapest_first = sorted(inventory, key=lambda x: x["price"])

print(cheapest_first)
````
