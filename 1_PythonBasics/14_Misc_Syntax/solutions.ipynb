## Exercise 1: A Basic Decorator

Write a decorator function called `print_before_after` that, when applied to a function, prints `"Before"` before calling the function and `"Print after"` after it returns. Apply it to a function `say_hello()` (which just prints `"Hello!"`) using `@` syntax, then call `say_hello()`.

```python
def print_before_after(func):
    def wrapper():
        print("Before")
        result = func()
        print("Print after")
        return result
    return wrapper


@print_before_after
def say_hello():
    print("Hello!")


say_hello()
```

## Exercise 2: Decorating a Function with Arguments

Write a decorator called `announce` that prints `"Calling function..."` before running the wrapped function and returns whatever the wrapped function returns. Apply it with `@` syntax to a function `add(a, b)` that returns `a + b`, so that the decorator works no matter what arguments `add` is called with. Call `add(3, 4)` and print the result.

```python
def announce(func):
    def wrapper(*args, **kwargs):
        print("Calling function...")
        result = func(*args, **kwargs)
        return result
    return wrapper


@announce
def add(a, b):
    return a + b


print(add(3, 4))
```

## Exercise 3: Reading and Writing a File with `with`

Using `with open(...) as ...:` syntax, write the string `"Practice makes perfect"` to a file called `practice.txt`. Then, in a separate `with` block, open `practice.txt` again for reading and print its contents.

```python
with open("practice.txt", "w") as f:
    f.write("Practice makes perfect")

with open("practice.txt", "r") as f:
    contents = f.read()

print(contents)
```

## Exercise 4: A Custom Context Manager Class

Write a class called `Announcer` that supports the `with` statement by defining `__enter__` and `__exit__`. `__enter__` should print `"Entering block"` and return `self`. `__exit__` should print `"Exiting block"`. Use it in a `with Announcer() as a:` block and print `"Inside block"` somewhere in the middle.

```python
class Announcer:
    def __enter__(self):
        print("Entering block")
        return self

    def __exit__(self, exc_type, exc_value, traceback):
        print("Exiting block")


with Announcer() as a:
    print("Inside block")
```

## Exercise 5: Walrus Operator in an `if` Statement

Given `numbers = [3, 7, 12, 5]`, use the walrus operator `:=` inside an `if` condition to assign `total = sum(numbers)` and, in the same condition, check whether `total` is greater than 20. If it is, print the value of `total`.

```python
numbers = [3, 7, 12, 5]

if (total := sum(numbers)) > 20:
    print(total)
```

## Exercise 6: Walrus Operator in a `while` Loop

Given `words = iter(["apple", "banana", "stop", "cherry"])`, write a `while` loop that uses the walrus operator to assign each value from `next(words)` to a variable `word` and continues looping as long as `word != "stop"`, printing each word along the way.

```python
words = iter(["apple", "banana", "stop", "cherry"])

while (word := next(words)) != "stop":
    print(word)
```

## Exercise 7: Walrus Operator in a List Comprehension

Given `values = [2, 4, 6, 8, 10]`, use a list comprehension with the walrus operator to build a list of the squares of each value, but only include a square in the result if it is greater than 20. (Use `:=` so the square is only computed once per value.)

```python
values = [2, 4, 6, 8, 10]

results = [square for x in values if (square := x * x) > 20]
print(results)
```

## Exercise 8: Combine Decorators, Context Managers, and the Walrus Operator

Write a decorator called `count_calls` that keeps track of how many times the decorated function has been called (store the count as an attribute on the wrapper function, e.g. `wrapper.calls`) and prints the current call count each time the function runs. Apply it with `@` syntax to a function `ping()` that just prints `"pong"`. Then, inside a `with Announcer() as a:` block (reusing your `Announcer` class from Exercise 4), call `ping()` three times using a `while` loop whose condition uses the walrus operator to assign and check a counter variable.

```python
def count_calls(func):
    def wrapper(*args, **kwargs):
        wrapper.calls += 1
        print("Call count:", wrapper.calls)
        return func(*args, **kwargs)
    wrapper.calls = 0
    return wrapper


@count_calls
def ping():
    print("pong")


with Announcer() as a:
    counter = 0
    while (counter := counter + 1) <= 3:
        ping()
```
