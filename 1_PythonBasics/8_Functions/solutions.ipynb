## Exercise 1 — `def` syntax with a default argument

Write a function `power(base, exponent=2)` that returns `base` raised to `exponent`. Call it once with only `base` supplied (relying on the default), and once supplying both arguments.

````python
def power(base, exponent=2):
    return base ** exponent


print(power(5))       # uses default exponent -> 25
print(power(2, 10))   # both arguments supplied -> 1024
````

## Exercise 2 — Positional vs. keyword arguments

Write a function `format_name(first, last, middle="")` that returns a full name string. Call it three different ways: (a) all positional, (b) all keyword arguments in a different order than the parameter list, (c) a mix of positional and keyword.

````python
def format_name(first, last, middle=""):
    if middle:
        return f"{first} {middle} {last}"
    return f"{first} {last}"


print(format_name("Ada", "Lovelace"))                               # (a) all positional
print(format_name(last="Lovelace", first="Ada", middle="Kate"))     # (b) all keyword, different order
print(format_name("Ada", last="Lovelace", middle="Kate"))           # (c) mix of positional and keyword
````

## Exercise 3 — `*args`

Write a function `join_words(*args)` that accepts any number of positional string arguments and returns them joined with a single space. Print the `type()` of `args` inside the function, then call it with 0, 1, and 4 arguments.

````python
def join_words(*args):
    print(type(args))
    return " ".join(args)


print(repr(join_words()))
print(repr(join_words("hello")))
print(repr(join_words("the", "quick", "brown", "fox")))
````

## Exercise 4 — `**kwargs`

Write a function `print_settings(**kwargs)` that accepts any number of keyword arguments and prints each `key = value` pair on its own line. Call it with at least three different keyword arguments (e.g. `debug=True, retries=3, name="job1"`).

````python
def print_settings(**kwargs):
    for key, value in kwargs.items():
        print(f"{key} = {value}")


print_settings(debug=True, retries=3, name="job1")
````

## Exercise 5 — Combining positional, `*args`, keyword-only, and `**kwargs`

Write a single function `make_request(method, *headers, timeout=30, **params)` (do not change this signature) that returns a dict summarizing its inputs, e.g. `{"method": method, "headers": headers, "timeout": timeout, "params": params}`. Call it in a way that exercises every part of the signature at once (a positional `method`, two or more extra positional `headers`, an explicit `timeout`, and two or more extra keyword arguments).

````python
def make_request(method, *headers, timeout=30, **params):
    return {
        "method": method,
        "headers": headers,
        "timeout": timeout,
        "params": params,
    }


result = make_request(
    "GET",
    "Authorization: Bearer token",
    "Accept: application/json",
    timeout=60,
    page=1,
    limit=25,
)
print(result)
````

## Exercise 6 — Unpacking at the call site

Given the function below (do not modify it) and the list/dict provided, call `create_event` **twice**: once unpacking `event_list` with `*` to supply the positional arguments, and once unpacking `event_dict` with `**` to supply the keyword arguments.

```python
def create_event(title, date, location="TBD"):
    return f"{title} on {date} at {location}"

event_list = ["Sprint Review", "2026-08-01"]
event_dict = {"title": "Retro", "date": "2026-08-02", "location": "Room 4"}
```

````python
def create_event(title, date, location="TBD"):
    return f"{title} on {date} at {location}"


event_list = ["Sprint Review", "2026-08-01"]
event_dict = {"title": "Retro", "date": "2026-08-02", "location": "Room 4"}

print(create_event(*event_list))
print(create_event(**event_dict))
````

## Exercise 7 — Lambdas with `sorted`, `map`, and `filter`

Given `records = [{"name": "Ada", "score": 88}, {"name": "Sam", "score": 95}, {"name": "Kai", "score": 72}]`:

1. Use `sorted(...)` with a `lambda` as the `key` to sort `records` by `"score"` descending.
2. Use `map(...)` with a `lambda` to produce a list of just the names, upper-cased.
3. Use `filter(...)` with a `lambda` to keep only records with `"score" >= 80`.

````python
records = [{"name": "Ada", "score": 88}, {"name": "Sam", "score": 95}, {"name": "Kai", "score": 72}]

# 1. sorted() with a lambda key, descending by score
by_score_desc = sorted(records, key=lambda record: record["score"], reverse=True)
print(by_score_desc)

# 2. map() with a lambda to get upper-cased names
names_upper = list(map(lambda record: record["name"].upper(), records))
print(names_upper)

# 3. filter() with a lambda to keep score >= 80
high_scorers = list(filter(lambda record: record["score"] >= 80, records))
print(high_scorers)
````

## Exercise 8 — Adding type hints

Add appropriate type hints to the function below without changing its behavior. The function should be annotated to show: `items` is a `list` of `int`, `factor` is an `int` with a default value, the return type is `list[int]`. Then write a second version, `find_first_negative`, that takes a `list[int]` and returns an `Optional[int]` (the first negative number found, or `None` if there isn't one) — you'll need to `import` whatever you use from `typing`.

```python
def scale(items, factor=2):
    return [item * factor for item in items]
```

````python
from typing import Optional


def scale(items: list[int], factor: int = 2) -> list[int]:
    return [item * factor for item in items]


def find_first_negative(items: list[int]) -> Optional[int]:
    for item in items:
        if item < 0:
            return item
    return None


print(scale([1, 2, 3]))
print(scale([1, 2, 3], factor=10))
print(find_first_negative([4, 7, -3, 9, -1]))
print(find_first_negative([4, 7, 9]))
````
